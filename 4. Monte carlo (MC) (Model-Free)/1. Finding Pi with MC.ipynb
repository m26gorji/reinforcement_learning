# **Finding Pi with MC**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 10000
Number_dot_inside = 0
x_red = []
y_red = []
x_blue = []
y_blue = []

for i in range(N):
  x = np.random.rand()
  y = np.random.rand()

  if np.sqrt(x**2 + y**2) < 1:
    Number_dot_inside = Number_dot_inside + 1
    x_red.append(x)
    y_red.append(y)
  else:
    x_blue.append(x)
    y_blue.append(y)


print('Number of dots inside the circle = ', Number_dot_inside)
print('Number of total dots = ', N)
print('Pi estimated ≈ ', 4*(Number_dot_inside / N))


plt.figure(figsize=(5, 5))
plt.scatter(x_red, y_red, color='red', s=1)
plt.scatter(x_blue, y_blue, color='blue', s=1)
plt.title(f"Indise's dot = {Number_dot_inside}, n = {N}, pi ≈ {4*(Number_dot_inside / N)}")
plt.axis('equal')
plt.show()
```

# **Value-state Function - Bellman Equation (Grid world 1D)**

Gamma = 0.5

V(a) = 1/2 (+1 + Gamma * V(b) + 2 + Gamma * V(c))

V(b) = 1/2 (+2 + Gamma * V(c) + (-1) + Gamma * V(a))

V(c) = 1/2 (-2 + Gamma * V(a) + (-2) + Gamma * V(b))

V(a) - 1/4 * V(b) - 1/4 * V(c) = 3/2

-1/4 * V(a) + V(b) - 1/4 * V(c) = 1/2

-1/4 * V(a) - 1/4 * V(b) + V(c) = -2

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([[1, -1/4,  -1/4], [ -1/4, 1,  -1/4], [ -1/4,  -1/4, 1]])
B = np.array([[3/2],[1/2],[-2]])
V = np.dot(np.linalg.inv(A), B)
print('A = ', A)
print('-------------------------------------')
print('B = ', B)
print('-------------------------------------')
print('V(a) = ', V[0][0])
print('V(b) = ', V[1][0])
print('V(c) = ', V[2][0])
```
